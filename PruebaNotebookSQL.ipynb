### 1. Lista los empleados por departamento

```sql
USE Empresa;

SELECT e.nss_Emp,apellidoPat_Emp,apellidoMat_Emp,nombres_Emp 
FROM Empleado e 
    JOIN Departamento d ON e.numero_Dep = d.numero_Dep 
WHERE d.nombre_Dep = 'Administración'
ORDER BY apellidoPat_Emp, apellidoMat_Emp, nombres_Emp;
```

### 2. Lista los empleados que no tienen dependientes 

```sql
USE Empresa;

SELECT e.nss_Emp,apellidoPat_Emp,apellidoMat_Emp,nombres_Emp
FROM Empleado e LEFT JOIN Dependiente d ON e.nss_Emp=d.nss_Emp
WHERE d.nss_Emp IS NULL; 
```

### 3. Lista la cantidad de empleados por departamento

```sql
SELECT nombre_Dep, COUNT(*) AS cantidad_empleados
FROM Departamento 
GROUP BY nombre_Dep;
```
