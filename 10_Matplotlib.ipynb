Matplotlib es una librería de Python para realizar representaciones gráficas de funciones, conjunto de datos, etc., utilizando diversos tipos de gráficos y diferentes fuentes de datos. Se caracteriza porque es fácil de usar, flexible y se puede configurar de múltiples maneras.  
Se trata de una de las herramientas clave en la tarea de cualquier análisis de datos, dado que además de procesar los datos, siempre será necesario poder visualizar los resultados de los procesamientos para poder tomar decisiones y obtener conclusiones.

Para importar Matplotlib desde cualquier programa de Python se hará de la siguiente manera:

```python
import matplotlib.pyplot as plt
import numpy as np
```

**1. La función Plot**

La función plot es la herramienta más básica de Matplotlib. Permite representar gráficamente una lista de pares de valores (x,y) sobre un eje de coordenadas uniendo cada punto de acuerdo al orden en que aparecen.

Representación de la lista de pares ordenados: (1,-3), (2,1), (3,4), (0,5)

```python
plt.plot([1,2,3,0],[-3,1,4,5])
plt.show()
```

Podemos cambiar el tamaño del gráfico estableciendo el valor del parámetro **figsize** en el método **figure**() durante la inicialización, que especifica el ancho y la altura de la figura en pulgadas.

```python
# Tiene que definirse antes que se dibuje el gráfico
plt.figure(figsize=(8,6))
plt.plot([1,2,3,0],[-3,1,4,5])

plt.show()
```

El valor predeterminado del parámetro **figsize** es [6.4, 4.8]

Si la figura ya está creada, podemos calcular el tamaño de la figura utilizando los métodos **set_figheight**() y **set_figwidth**().

```python
# Se tiene que definir una variable figura (fig)
fig=plt.figure()

plt.plot([1,2,3,0],[-3,1,4,5])

fig.set_figheight(6) # altura
fig.set_figwidth(8) # ancho

plt.show()
```

Otra ópcion es usar el método **set_size_inches()** que también podría establecer el tamaño de la figura después de que se haya creado la figura.

```python
# Se tiene que definir una variable figura (fig)
fig=plt.figure()

plt.plot([1,2,3,0],[-3,1,4,5])

fig.set_size_inches(8,6)

plt.show()
```

Aquí los argumentos pasados a​​l método **set_size_inches()** representan el ancho y la altura de la figura en pulgadas respectivamente.

Se puede omitir la lista de valores para el eje x, y en este caso, se usa como valores por defecto del eje x, la lista de valores que van desde 0 (primer valor) hasta *n-1* donde *n* es el número de valores proporcionados para el eje y.

**2. Elementos decorativos de un gráfico**

**2.1. Eje de coordenadas**

Matplotlib por defecto establece los valores que aparecen en los ejes de coordenadas de manera que puedan mostrarse en la gráfica los puntos que son dibujados  
Para configurar esos valores se usa la función **axis**. Cuando se ejecuta sin parámetros devuelve la escala actual utilizada en los ejes en forma de una tupla de 4 valores que indican el límite inferior y superior de la escala de valores usada para el eje x y ´para el eje y respectivamente. Ver la siguiente figura:

```python
import numpy as np

x = np.arange(8)
plt.plot(x,[xi**2 for xi in x])
plt.axis()

#plt.show()
```

Para cambiar las escalas usadas se invoca la función **axis** mediante una lista de 4 valores que representan el valor mínimo del eje x, el valor máximo del eje x, el valor mínimo del eje y, y el valor máximo del eje y.

```python
x = np.arange(8)

plt.plot(x,[xi**2 for xi in x])
plt.axis([1,8,-3,8])
plt.show()
```

Analogamente, se podria haber invocado la función **axis** usando el formato de los argumentos *xmin*, *xmax*, *ymin* y *ymax*.

```python
x = np.arange(8)

plt.plot(x,[xi**2 for xi in x])
plt.axis(xmin=1,xmax=8,ymin=-3,ymax=8)
plt.show()
```

La función axis se puede invocar para configurar solo uno de los ejes. Para ello habrá que proporcionarle los valores correspondientes de la escala utilizando la notación de parámetros argumentales.  
Por ejemplo, si se quiere modificar sólo el eje y de manera que varíe entre 4 y 10, se le invocaría de la siguiente forma:

```python
x = np.arange(8)

plt.plot(x,[xi**2 for xi in x])
plt.axis(ymin=4,ymax=10)
plt.show()
```

Existen dos funciones  **xlim** e **ylim** que permiten controlar las escalas de los ejes x e y respectivamente de manera independiente, así:
*   Si se invoca sin parámetros, devuelve los valores actuales de las escalas de los ejes x e y.
*   Y si se invoca con una lista de 2 valores, entonces se modifican las escalas de los ejes x e y. La lista de valores representa el límite inferior y superior de la escala de un eje, respectivamente. En este caso, las funciones pueden ser invocadas en forma de lista o bien con parámetros argumentales.



```python
x = np.arange(8)

plt.plot(x,[xi**2 for xi in x])
print(plt.xlim())
print(plt.ylim())
plt.show()
```

```python
x = np.arange(8)

plt.plot(x,[xi**2 for xi in x])
plt.xlim([1,8])
plt.ylim(ymin=-3, ymax=8)
plt.show()
```

Para gestionar las etiquetas que se colocan sobre los ejes y la localización de ellas, se usan  las funciones **xticks** y **yticks** que permiten manipular estos elementos en los ejes x e y, respectivamente.   
Para ambos casos toman como parámetros dos listas de la misma longitud que contienen las localizaciones de los valores en los ejes, y por otra parte, los valores que se desean colocar en esas localizaciones.  
Esta ultima lista se puede omitir, de manera que se tomaría como valores, los usados en las localizaciones.

```python
x = range(8)

plt.plot(x,[xi for xi in x])

plt.xticks(range(8),["a","b","c","d","e","f","g","h"])
plt.yticks(range(0,16,2))
plt.show()
```

**2.5. Leyenda**

Para añadir una leyenda a la representación de un conjunto de puntos se invoca la función **plot** que dibuja la secuencia con el argumento **label** al cual se le asigna el valor que debe tomar la leyenda (esta asignación debe aparecer a continuación de los parámetros que aparecen sin formato de argumentos). Por último, se invoca la función legend sin argumentos para que dibuje las leyendas.

```python
x = range(8)

plt.plot(x, [xi for xi in x], label="Lineal")
plt.plot(x, [xi**2 for xi in x], label="Cuadrática")
plt.plot(x, [xi**3 for xi in x], label="Cúbica")
plt.legend()
plt.show()
```

Otra forma

```python
x = range(8)

plt.plot(x, [xi for xi in x])
plt.plot(x, [xi**2 for xi in x])
plt.plot(x, [xi**3 for xi in x])
plt.legend(["Lineal","Cuadrática","Cúbica"])
plt.show()
```

Se puede indicar lugar del gráfico en el cual se dibujará la leyenda añadiendo el argumento **loc** a la funcion **legend**. Este argumento se toma como valor una cadena o un número tal como aparece en el gráfico siguiente:

```python
x = range(8)

plt.plot(x, [xi for xi in x])
plt.plot(x, [xi**2 for xi in x])
plt.plot(x, [xi**3 for xi in x])
#plt.legend(["Lineal","Cuadrática","Cúbica"], loc="center")
plt.legend(["Lineal","Cuadrática","Cúbica"], loc=10)
plt.show()
```

Argumentos de la función legend():  
0: best  
1: upper right  
2: upper left  
3: lower right  
4: lower left  
5: right  
6: center left  
7: center right  
8: lower center  
9: upper center  
10: center

También es posible especificar la situación de la leyenda mediante un par de números que indican las coordenadas (x,y) de la esquina inferior izquierda de la leyeda donde se quiere situar dentro de la gráfica. Así por ejemplo (0,1) indica la parte superior izquierda, (0.5, 0.5) indica el centro, etc.

```python
x = range(8)

plt.plot(x, [xi for xi in x])
plt.plot(x, [xi**2 for xi in x])
plt.plot(x, [xi**3 for xi in x])
plt.legend(["Lineal","Cuadrática","Cúbica"], loc=(0,0)) # (0.5,0.5) indica centro y (0,1) indica la parte superior izquierda
plt.show()
```

**ESTILO Y ANCHO DE LINEA**

**linestyle**  
Permite especificar -como indica su nombre- el estilo de línea. Por defecto es una línea continua que une los diferentes puntos, pero podría ser, por ejemplo una línea discontinua:

**Estilo:**  
\- Línea sólida  
--  Línea discontinua  
-.  Línea discontinua con puntos  
:  Línea de puntos

```python
# Se genera 50 números aleatorios en el rango [0, 1) extraídos de una distribución uniforme, y sumados de forma acumulativa
y = np.random.randn(50).cumsum()

plt.plot(y)
plt.show()
```

```python
plt.plot(y, linestyle="--")
plt.show()
```

```python
a = np.random.randn(50).cumsum()
b = np.random.randn(50).cumsum()
c = np.random.randn(50).cumsum()
d = np.random.randn(50).cumsum()
e = np.random.randn(50).cumsum()


plt.plot(a, linestyle="-", label="solid")
plt.plot(b, linestyle="--", label="dashed")
plt.plot(c, linestyle="-.", label="dashdot")
plt.plot(d, linestyle=":", label="dotted")
plt.plot(e, linestyle="None", label="None")

plt.legend()

plt.show()
```

```python
x = range(1,11)
y = [1]

plt.plot(x,[1]*10, marker=".", label=".")
plt.plot(x,[2]*10, marker="o", label="o")
plt.plot(x,[3]*10, marker="v", label="v")
plt.plot(x,[4]*10, marker="1", label="1")
plt.plot(x,[5]*10, marker="s", label="s")
plt.plot(x,[6]*10, marker="p", label="p")
plt.plot(x,[7]*10, marker="P", label="P")
plt.plot(x,[8]*10, marker="*", label="*")
plt.plot(x,[9]*10, marker="+", label="+")
plt.plot(x,[10]*10, marker="d", label="d")
plt.plot(x,[11]*10, marker="D", label="D")

plt.legend(loc=(1.1,0))

plt.show()
```

Junto a este parámetro hay otros que permiten ajustar el ancho del borde del marcador (**markeredgewidth**), el color del borde del marcador (**markeredgecolor**), el color de fondo del marcador (**markerfacecolor**), el color alternativo para el color de fondo del marcador (**markerfacecoloralt**) y el tamaño del marcador (**markersize**). Veamos este último:

**markersize**  
Este parámetro acepta un número real. Apliquémoslo con el marker "o"

```python
y = np.random.randn(50).cumsum()

plt.plot(y, marker="o", markersize=12)
plt.show()
```

```python
x = range(8)

plt.plot(x, [xi for xi in x], marker="d", label="Lineal", markersize=10, markerfacecolor="y", markeredgecolor="m")
plt.plot(x, [xi**2 for xi in x], marker="p", label="Cuadrática", markersize=8, markerfacecolor="c", markeredgecolor="g")
plt.plot(x, [xi**3 for xi in x], marker="s", label="Cúbica", markersize=5, markerfacecolor="r", markeredgecolor="b")
plt.legend()
plt.show()
```

Podemos usar también los argumentos:  
- **ms:** tamaño del marcador (markersize)
- **mfc:** color de relleno del marcador (markerfacecolor)
- **mec:** color del borde del marcador (markeredgecolor)

```python
x = range(8)

plt.plot(x, [xi for xi in x], marker="d", label="Lineal",ms=10, mfc="y", mec="m")
plt.plot(x, [xi**2 for xi in x], marker="p", label="Cuadrática",ms=8, mfc="c", mec="g")
plt.plot(x, [xi**3 for xi in x], marker="s", label="Cúbica",ms=5, mfc="r", mec="b")
plt.legend()
plt.show()
```

```python
x = range(8)

plt.plot(x, [xi for xi in x], marker="D", linestyle=":", color="r", label="Lineal",ms=10, mfc="y", mec="m")
plt.plot(x, [xi**2 for xi in x], marker="p", linestyle="--", color="b", label="Cuadrática",ms=8, mfc="c", mec="g")
plt.plot(x, [xi**3 for xi in x], marker="s", linestyle="-", color="k", label="Cúbica",ms=5, mfc="r", mec="b")
plt.legend()
plt.show()
```

**color**  
Este parámetro controla el color de la gráfica y acepta gran cantidad de formatos distintos:

- Uno de los siguientes nombres: 'blue', 'orange', 'green', 'red', 'purple', 'brown', 'pink', 'gray', 'olive' o 'cyan'

- Un color en formato X11/CSS4  
- Un nombre de color del xkcd color survey precedido por "xkcd:" (por ejemplo, "xkcd:burnt orange")  
- Cualquier nombre de color web, por ejemplo "Tomato" o "CornflowerBlue"

Observar que se puede configurar el formato para que los puntos no sean unidos mediante segmentos. Para ello basta no indicar nada en la cadena de formato acerca del color y el estilo de las líneas, de manera que solo se dibujarán los puntos.

```python
x = range(8)

plt.plot(x, [xi for xi in x], "d", label="Lineal")
plt.plot(x, [xi**2 for xi in x], "p", label="Cuadrática")
plt.plot(x, [xi**3 for xi in x], "s", label="Cúbica")
plt.legend()
plt.show()
```

Como se ha comentado anteriormente, los colores representados por una letra ( 'b', 'g', 'r', 'c', 'm', 'y', 'k', 'w') son los menos atractivos de la paleta, por lo que la opción de fijar el color de la gráfica mediante este parámetro [fmt] puede ser útil para trabajos propios o pruebas de concepto, siendo más recomendable fijar colores más agradables mediante el parámetro color cuando el informe se destine a un tercero

```python
y1 = np.random.randn(50).cumsum()
y2 = np.random.randn(50).cumsum()

plt.plot(y1, "-g", y2, "-.*r")
plt.show()
```

**3.5. Diagramas circulares**

Se trata de un gráfico en forma de círculo que se encuentra dividido en sectores, de manera que la superficie ocupada por el sector representa proporcionalmente a la cantidad descrita.



Para representar gráfico circulares se utiliza la función **plt.pie** que toma como entrada un array X con valores. Así para cada valor x del array X, la función genera sectores que son proporcionales a **x / Suma(X)**. Si la suma fuera menor a 1, entonces se representan los valores directamente. Los sectores se dibujan empezando desde el eje horizontal, en el lado derecho y en sentido contrario a las manecillas del reloj.

La función tienen los siguintes parámetros:   
- **explode**: es un array de la misma longitud que el array de valores X, cuyos valores indican la fracción del radio que va a separar el sector del centro del círculo.
- **colors**: es la lista de colores que se usarán cíclicamente para los sectores. En caso de no indicar, sigue una prograsión automática de: azul, verde, rojo, cian, magenta, ...   
- **labels**: es una lista de las etiquetas asociadas a cada sector
- **labeldistance**: es la distancia respecto al centro del gráfico en la que se dibuja una etiqueta.

```python
import numpy as np
import matplotlib.pyplot as plt
dx=0.01;dy=0.01
x=np.arange(-2.0,2.0,dx)
y=np.arange(-2.0,2.0,dy)
X,Y=np.meshgrid(x,y)
def f(x,y):
    return(1-y**5+x**5)*np.exp(-x**2-y**2)

C = plt.contour(X, Y, f (X, Y), 8, colors = 'black') #Generar mapa de contorno
plt.contourf(X,Y,f(X,Y),8)
plt.clabel(C,inline=1,fontsize=10)
plt.colorbar()
plt.show()
```

Una gráfica de caja y bigotes generalmente se usa para resumir la distribución de una muestra de datos siendo una alternativa a los histogramas.

Las líneas llamadas bigotes se dibujan extendiéndose desde ambos extremos de la caja con la longitud del IQR para demostrar el rango esperado de valores sensibles en la distribución. Los cuantiles son puntos tomados a intervalos regulares de la función de distribución de una variable aleatoria.

Estos diagramas pueden ayudar a obtener rápidamente una idea del rango de valores comunes y sensibles en la caja y en el bigote, respectivamente. **4. SUBGRÁFICOS**

En Matplotlib existe a posibilidad de crear gráficos formados por un conjunto de gráficos. A estos gráficos se les denomina *subgráficos*.  
Para ello se utiliza la función **plt.figure**, que genera un objeto de tipo **Figure**, que es un contenedor donde se pueden añadir subgráficos utilizando el método **add_subplot**. La invocación del método devuelve un objeto de tipo **axe**, que es un área donde se puede dibujar.

El método **add_subplot** asociado a una figura añade un conjunto de ejes a la misma pero suponiéndolo en una matriz de columnas y filas (de conjuntos de ejes) y creándolo en la posición indicada de dicha matriz

Los parámetros del método permiten especificar de manera matricial donde se insertan los subgráficos:  
- **numrows:** especifica el número de filas  
- **numcols**: especifica el número de columnas  
- **fignum**: especifica un número que varía entre 1 y **numrows x numcols**, que indica el subgráfico actual, avanzando de izquierda a derecha y de arriba abajo. Así, 1 indica la esquina superior izquierda y **numrows x numcols** indica la esquina inferior derecha.

Los objetos de tipo **axe** pueden invocar la funcion **plt.plot** para representar puntos, como se muestra en el siguiente ejemplo:

Si se invoca el método sin añadir argumentos, se supone que dicha matriz está formada únicamente por un conjunto de ejes (es decir, una columna y una fila):

```python
import numpy as np
# cumsum devuelve la suma acumulada de los valores de un array a lo largo de los ejes especificados


y = np.random.randn(100).cumsum()

plt.plot(y)
plt.show()
```

```python
y = np.random.randn(100).cumsum()

fig = plt.figure()
fig.add_subplot()
plt.plot(y)
plt.show()
```

Pero si se invoca añadiendo como argumentos el número de filas de la matriz, el número de columnas de la matriz y la posición que ocupa el conjunto de ejes a crear, se supone una matriz de dichas dimensiones y se sitúa el conjunto de ejes en la posición correcta:



```python
import matplotlib.pyplot as plt

fig = plt.figure()

x = range(8)

#Se indica dos filas y una columna. Es decir se forma una matriz de 2 x 1
#Se asigna 1 para el primer gráfico, y 2 para el segundo gráfico
# numrows = 2, numcols = 1
ax1 = fig.add_subplot(2,1,1)
ax1.plot(x, [xi for xi in x])


ax2 = fig.add_subplot(2,1,2)
ax2.plot(x, [xi**2 for xi in x])

plt.show()
```

Dos parámetros interesantes de la función **subplots** son **sharex** y **sharey**. Éstos controlan la compartición de las propiedades de los ejes.
```python
fig, ax = plt.subplots(2,2, sharex=True)
fig.set_size_inches(8,4)

ax[1,0].plot(y)
plt.show()
```

En algunas situaciones puede ser necesario dibujar varios **gráficos superpuestos** que representan diferentes conjuntos de datos compartiendo el eje x o bien el eje y.  
En cualquiera de los casos, lo primero será crear un objeto de tipo **axe** (que contendrá uno de los conjuntos de datos). A continuación, se utilizará la función **twinx** (para compartir el eje x) o **twiny** (para compartir el eje y) que permite crear un nuevo objeto **axe** superpuesto (estas funciones se pueden invocar desde objetos de tipo axe y devuelven objetos de tipo axe).  
En los gráfico superpuestos se utilizarán diferentes escalas para los ejes x o y dependiendo del eje que comparten (ver figura), como puede verse en el ejemplo siguiente:

```python
x = range(8)

fig = plt.figure()

ax1 = fig.add_subplot(1,1,1)

ax1.plot(x, [xi for xi in x])
ax1.set_ylabel("Función lineal")

ax2 = ax1.twinx() # compartir el eje x
ax2.plot(x, [xi**2 for xi in x], "r")
ax2.set_ylabel("Función cuadrática")

ax1.set_xlabel("Eje x compartido")

plt.show()
```

```python
x = range(8)

fig = plt.figure()

ax1 = fig.add_subplot(1,1,1)

ax1.plot(x, [xi for xi in x])
ax1.set_ylabel("Función lineal")

ax2 = ax1.twiny() # compartir el eje y
ax2.plot(x, [xi**2 for xi in x], "r")
ax2.set_ylabel("Función cuadrática")

ax1.set_xlabel("Eje x compartido")

plt.show()
```
